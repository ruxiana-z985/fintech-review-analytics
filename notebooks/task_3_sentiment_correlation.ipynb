# Task 3: Sentiment and Correlation Analysis

This notebook scores headline sentiment, aligns news to trading days, computes daily returns, and measures Pearson correlation.

```python
from src.data_loader import load_news_data, copy_price_data, load_stock_prices
from src.sentiment_analysis import add_sentiment_scores
from src.correlation_analysis import align_news_to_trading_day, aggregate_daily_sentiment, compute_daily_returns, combine_sentiment_and_returns, correlation_by_stock

copy_price_data()
news_df = load_news_data()
news_df = add_sentiment_scores(news_df)
aapl_prices = load_stock_prices('AAPL')
aapl_news = news_df[news_df['stock'] == 'AAPL']
aligned = align_news_to_trading_day(aapl_news, aapl_prices)
daily_sentiment = aggregate_daily_sentiment(aligned)
daily_returns = compute_daily_returns(aapl_prices)
combined = combine_sentiment_and_returns(daily_sentiment, daily_returns)
combined.head()
```

```python
correlation_by_stock(combined.assign(stock='AAPL'))
```
